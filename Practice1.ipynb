```python
in_vowels="aeiou"
out_stars="*****"
delete_punc="!,."
table=str.maketrans(in_vowels,out_stars,delete_punc)

txt="Hello World!!"
print(txt.translate(table))
```

```python
mapping = {"a": "4", "e": "3", "i": None}
table = str.maketrans(mapping)

# Apply to a string
text = "apple juice"
print(text.translate(table))
```

```python
# Swap 'a', 'b', 'c' with 'x', 'y', 'z'
in_chars = "abc"
out_chars = "xyz"
table = str.maketrans(in_chars, out_chars)

text = "cat and bat"
print(text.translate(table))  # Output: zxt xnd yxt

```
